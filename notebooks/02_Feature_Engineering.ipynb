```python
import pandas as pd
from pathlib import Path

# Setup paths
project_root = Path.cwd().parent
raw_dir = project_root / "data" / "raw"
sample_dir = project_root / "data" / "sample"

# Ensure the sample directory exists
sample_dir.mkdir(parents=True, exist_ok=True)

# 1. Create a sample of Benign traffic
try:
    # We only read the first 100 rows to save memory and time
    benign_sample = pd.read_csv(raw_dir / "benign.csv", nrows=100)
    benign_sample.to_csv(sample_dir / "sample_benign.csv", index=False)
    print("✅ Successfully created data/sample/sample_benign.csv")
except FileNotFoundError:
    print("❌ Could not find benign.csv in data/raw/")

# 2. Create a sample of Attack traffic (using cobaltstrike as an example)
try:
    attack_sample = pd.read_csv(raw_dir / "attacks" / "cobaltstrike.csv", nrows=100)
    attack_sample.to_csv(sample_dir / "sample_cobaltstrike.csv", index=False)
    print("✅ Successfully created data/sample/sample_cobaltstrike.csv")
except FileNotFoundError:
    print("❌ Could not find cobaltstrike.csv in data/raw/attacks/")
```
